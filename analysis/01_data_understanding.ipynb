```python
branch_analysis = (
    df_clean.groupby("branch")
    .agg(
        total_students=("student_id", "count"),
        placed_students=("placed", "sum"),
        placement_rate=("placed", "mean")
    )
)

branch_analysis["placement_rate"] *= 100

branch_analysis = branch_analysis.sort_values(
    "placement_rate",
    ascending=False
)

branch_analysis
```

```python

df_clean["cgpa_group"] = pd.cut(
    df_clean["cgpa"],
    bins=[0, 6, 7, 8, 9, 10],
    labels=["Below 6", "6-7", "7-8", "8-9", "9-10"],
    include_lowest=True
)

cgpa_analysis = (
    df_clean.groupby("cgpa_group", observed=True)
    .agg(
        total_students=("student_id", "count"),
        placed_students=("placed", "sum"),
        placement_rate=("placed", "mean")
    )
)

cgpa_analysis["placement_rate"] *= 100

cgpa_analysis
```

```python
# Communication skills and aptitude score vs placement

def placement_analysis(data, column):
    result = (
        data.groupby(column)
        .agg(
            total_students=("student_id", "count"),
            placed_students=("placed", "sum"),
            placement_rate=("placed", "mean")
        )
    )
    
    result["placement_rate"] *= 100
    return result.sort_index().round(2)


communication_analysis = placement_analysis(
    df_clean,
    "communication_skills"
)

aptitude_analysis = placement_analysis(
    df_clean,
    "aptitude_score"
)

print("=== COMMUNICATION SKILLS ===")
display(communication_analysis)

print("\n=== APTITUDE SCORE ===")
display(aptitude_analysis)
```

```python
# Group aptitude scores into meaningful ranges

df_clean["aptitude_group"] = pd.cut(
    df_clean["aptitude_score"],
    bins=[0, 40, 50, 60, 70, 80, 90, 100],
    labels=[
        "0-40",
        "41-50",
        "51-60",
        "61-70",
        "71-80",
        "81-90",
        "91-100"
    ],
    include_lowest=True
)

aptitude_group_analysis = (
    df_clean.groupby("aptitude_group", observed=True)
    .agg(
        total_students=("student_id", "count"),
        placed_students=("placed", "sum"),
        placement_rate=("placed", "mean")
    )
)

aptitude_group_analysis["placement_rate"] *= 100

aptitude_group_analysis.round(2)
```
